<h3>Dataset Tables & Descriptions:</h3>

| Table Name | Description | Key Attributes |
| :--- | :--- | :--- |
| `customers` | Customer demographics & unique identifier | `customer_id`, `customer_unique_id`, `customer_city`, `customer_state` |
| `orders` | Core order lifecycle & status milestones | `order_id`, `customer_id`, `order_status`, `order_purchase_timestamp` |
| `order_items` | Items inside each order with prices & freight | `order_id`, `order_item_id`, `product_id`, `seller_id`, `price` |
| `order_payments` | Payment methods & installment breakdowns | `order_id`, `payment_type`, `payment_installments`, `payment_value` |
| `order_reviews` | Customer satisfaction review ratings | `review_id`, `order_id`, `review_score`, `review_creation_date` |
| `products` | Product dimensions, weight & category names | `product_id`, `product_category_name`, `product_weight_g` |
| `sellers` | Seller locations across Brazil | `seller_id`, `seller_city`, `seller_state` |
| `geolocation` | Lat/Long geographic coordinates by zip code | `geolocation_zip_code_prefix`, `geolocation_lat`, `geolocation_lng` |
| `category_translation`| Translation of Portuguese category names to English | `product_category_name`, `product_category_name_english` |


```python
# =============================================================================
# DATA EXTRACTION & ENVIRONMENT SETUP
# This cell automatically loads all 9 CSV tables into Pandas DataFrames
# and registers them into an in-memory SQLite database ('conn') for SQL querying.
# =============================================================================

import os
import sqlite3
import pandas as pd
import numpy as np

# 1. Resolve Dataset Directory
DATA_DIR = os.path.join('data', 'olist')

if not os.path.exists(DATA_DIR) or not os.listdir(DATA_DIR):
    try:
        import kagglehub
        print('Downloading Olist Brazilian E-Commerce dataset via kagglehub...')
        DATA_DIR = kagglehub.dataset_download('olistbr/brazilian-ecommerce')
        print(f'Downloaded to: {DATA_DIR}')
    except Exception as err:
        print(f'Please ensure Olist CSV files are present in data/olist/: {err}')

print(f'Reading CSV files from: {os.path.abspath(DATA_DIR)}\n')

# 2. Extract DataFrames
customers = pd.read_csv(os.path.join(DATA_DIR, 'olist_customers_dataset.csv'))
geolocation = pd.read_csv(os.path.join(DATA_DIR, 'olist_geolocation_dataset.csv'))
orders = pd.read_csv(os.path.join(DATA_DIR, 'olist_orders_dataset.csv'))
order_items = pd.read_csv(os.path.join(DATA_DIR, 'olist_order_items_dataset.csv'))
order_payments = pd.read_csv(os.path.join(DATA_DIR, 'olist_order_payments_dataset.csv'))
order_reviews = pd.read_csv(os.path.join(DATA_DIR, 'olist_order_reviews_dataset.csv'))
products = pd.read_csv(os.path.join(DATA_DIR, 'olist_products_dataset.csv'))
sellers = pd.read_csv(os.path.join(DATA_DIR, 'olist_sellers_dataset.csv'))
category_translation = pd.read_csv(os.path.join(DATA_DIR, 'product_category_name_translation.csv'))

dfs = {
    'customers': customers,
    'geolocation': geolocation,
    'orders': orders,
    'order_items': order_items,
    'order_payments': order_payments,
    'order_reviews': order_reviews,
    'products': products,
    'sellers': sellers,
    'category_translation': category_translation
}

# 3. Register DataFrames into SQLite Database for instant SQL querying
conn = sqlite3.connect(':memory:')

print('=' * 65)
print(f"{'Table Name':<24} | {'Row Count':<14} | {'Columns':<8}")
print('=' * 65)
for name, df in dfs.items():
    df.to_sql(name, conn, if_exists='replace', index=False)
    print(f"{name:<24} | {len(df):<14,d} | {len(df.columns):<8}")
print('=' * 65)

# 4. Helper function to execute SQL queries
def run_query(sql_query: str) -> pd.DataFrame:
    """Executes a SQL query against the Olist database and returns a formatted DataFrame."""
    return pd.read_sql_query(sql_query, conn)

print('\n✅ All 9 DataFrames extracted and registered into SQL engine! Ready for queries.')
```
